# 01 Data Profiling
This notebook inspects the raw Bike Share Toronto ridership data and Toronto hourly weather data.
The goal is to understand the structure, missing values, duplicates, data types, and potential cleaning issues before transforming the data.

```python
import pandas as pd
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

project_root = Path("..")

bike_path = project_root / "data" / "raw" / "bikeshare-ridership-2026.csv"

weather_folder = project_root / "data" / "raw" / "weather"

weather_files = [
    weather_folder / "en_climate_hourly_ON_6158355_01-2026_P1H.csv",
    weather_folder / "en_climate_hourly_ON_6158355_02-2026_P1H.csv",
    weather_folder / "en_climate_hourly_ON_6158355_03-2026_P1H.csv",
]

print("Bike file exists:", bike_path.exists())

for file in weather_files:
    print(file.name, "exists:", file.exists())
```

```python
trips = pd.read_csv(bike_path)

weather_dfs = []

for file in weather_files:
    df = pd.read_csv(file)
    weather_dfs.append(df)

weather = pd.concat(weather_dfs, ignore_index=True)

print("Trips shape:", trips.shape)
print("Weather shape:", weather.shape)
```

```python
display(trips.head())
display(weather.head())
```

```python
print("Bike Share Data Types:")
print(trips.dtypes)

print("\nWeather Data Types:")
print(weather.dtypes)
```

```python
missing_trips = trips.isna().sum().sort_values(ascending=False)
missing_weather = weather.isna().sum().sort_values(ascending=False)

print("Missing values in Bike Share data:")
print(missing_trips)

print("\nMissing values in Weather data:")
print(missing_weather)
```

```python
print("Duplicate full rows in trips:", trips.duplicated().sum())
print("Duplicate Trip_Id values:", trips["Trip_Id"].duplicated().sum())

print("Duplicate full rows in weather:", weather.duplicated().sum())
print("Duplicate weather timestamps:", weather["Date/Time (LST)"].duplicated().sum())
```

```python
print(trips["Trip_Duration"].head(20))
print("\nTrip duration summary:")
print(trips["Trip_Duration"].describe())
```

```python
print("User types:")
print(trips["User_Type"].value_counts(dropna=False))

print("\nBike models:")
print(trips["Bike_Model"].value_counts(dropna=False))

print("\nNumber of start stations:", trips["Start_Station_Id"].nunique())
print("Number of end stations:", trips["End_Station_Id"].nunique())
print("Number of start station names:", trips["Start_Station_Name"].nunique())
print("Number of end station names:", trips["End_Station_Name"].nunique())
```

```python
trips["Start_Time_temp"] = pd.to_datetime(trips["Start_Time"], errors="coerce")
trips["End_Time_temp"] = pd.to_datetime(trips["End_Time"], errors="coerce")
weather["Date_Time_temp"] = pd.to_datetime(weather["Date/Time (LST)"], errors="coerce")

print("Bike trip start date range:")
print(trips["Start_Time_temp"].min(), "to", trips["Start_Time_temp"].max())

print("\nBike trip end date range:")
print(trips["End_Time_temp"].min(), "to", trips["End_Time_temp"].max())

print("\nWeather date range:")
print(weather["Date_Time_temp"].min(), "to", weather["Date_Time_temp"].max())

print("\nInvalid Start_Time values:", trips["Start_Time_temp"].isna().sum())
print("Invalid End_Time values:", trips["End_Time_temp"].isna().sum())
print("Invalid Weather Date/Time values:", weather["Date_Time_temp"].isna().sum())
```

```python
trips_jan_mar = trips[
    (trips["Start_Time_temp"] >= "2026-01-01") &
    (trips["Start_Time_temp"] < "2026-04-01")
].copy()

print("All trips:", trips.shape)
print("Jan-Mar trips:", trips_jan_mar.shape)
print("Percentage of trips in Jan-Mar:", round(len(trips_jan_mar) / len(trips) * 100, 2), "%")
```

monthly_trips = (
    trips_jan_mar
    .assign(month=trips_jan_mar["Start_Time_temp"].dt.to_period("M"))
    .groupby("month")
    .size()
    .reset_index(name="trip_count")
)

monthly_trips

```python
hourly_trips = (
    trips_jan_mar
    .assign(hour=trips_jan_mar["Start_Time_temp"].dt.hour)
    .groupby("hour")
    .size()
    .reset_index(name="trip_count")
)

hourly_trips
```
